# Example Usage – Diabetes Prediction Random Forest

This notebook demonstrates how to load and use the published Random Forest model
from Kaggle Model Hub for diabetes risk prediction using a recall-optimized
decision threshold.


```python
import joblib
import pandas as pd

MODEL_PATH = "/kaggle/input/diabetes-prediction-random-forest/scikitlearn/default/1/diabetes_rf_pipeline_v1.joblib"

artifact = joblib.load(MODEL_PATH)

model = artifact["model"]
threshold = artifact["threshold"]

sample = pd.DataFrame([{
    "Pregnancies": 2,
    "Glucose": 140,
    "BloodPressure": 80,
    "SkinThickness": 25,
    "Insulin": 100,
    "BMI": 30.5,
    "DiabetesPedigreeFunction": 0.5,
    "Age": 45
}])

prob = model.predict_proba(sample)[:, 1][0]
prediction = int(prob >= threshold)

print("Diabetes Probability:", round(prob, 3))
print("Prediction:", "Diabetes" if prediction else "No Diabetes")
```
